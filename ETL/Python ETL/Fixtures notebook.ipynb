```python
import pandas as pd
import constants
```

```python
# Data frames
website_frame = pd.read_csv(constants.CSV_SOURCE_URL)
fixtures_frame = pd.read_csv("../../dataset/2024-2025 fixtures.csv")
```

```python
fixtures_frame.head()
```

```python
import numpy as np

fixture_teams = fixtures_frame["Home Team"].unique()
website_teams = website_frame["HomeTeam"].unique()

# Teams in fixture_teams but not in website_teams
diff_fixture_to_website = np.setdiff1d(fixture_teams, website_teams)

# Teams in website_teams but not in fixture_teams
diff_website_to_fixture = np.setdiff1d(website_teams, fixture_teams)

print("In fixture_teams but not in website_teams:", diff_fixture_to_website)
print("In website_teams but not in fixture_teams:", diff_website_to_fixture)
```

```python
# Standardize names from website to fixtures for both home and away teams
mapping = {'Man Utd': 'Man United', 'Spurs': 'Tottenham'}
fixtures_frame["Home Team"] = fixtures_frame["Home Team"].replace(mapping)
fixtures_frame["Away Team"] = fixtures_frame["Away Team"].replace(mapping)
```

```python
fixtures_frame.insert(0, "MatchID", constants.CURRENT_SEASON_TAG + "_" + fixtures_frame["Home Team"] + "_" + fixtures_frame["Away Team"])
```

```python
fixtures_frame = fixtures_frame[["MatchID", "Round Number", "Date"]]
```

```python
fixtures_frame.to_csv("../../dataset/2024-2025 fixtures processed.csv", index=False)
# Save the updated fixtures frame  
```
